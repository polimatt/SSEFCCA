```python
def get_rounding(x:float):
    if x < 1 and x != 0:
        string = str(x)
        string = string.split('.')

        for i in range(len(string[1])):
            if i == len(string[1])-1:
                return i + 1

            elif string[1][i] != '0':
                if int(string[1][i+1]) >= 5 and string[1][i]=='9':
                    return i
                else: return i + 1
    else:
        return -len(str(int(np.round(x,0))))+1

def get_right_dec_figs_in_string(n,dec_places):
    dec_figs = len(str(n).split('.')[1])
    if dec_figs < dec_places:
        return str(n) + ''.join(['0']*(dec_places-dec_figs))
    else:
        return str(n)

def plus_minus(mean,std):
    std_dec_places = [get_rounding(x) if x!=0 else 0 for x in std]

    list_of_strings = []

    for x, y, z in zip(mean,std,std_dec_places):
        if y < 1 and y != 0:
            list_of_strings.append(f'{get_right_dec_figs_in_string(np.round(x,z),z)} ± {get_right_dec_figs_in_string(np.round(y,z),z)}')
        else:
            list_of_strings.append(f'{int(np.round(x,z))} ± {int(np.round(y,z))}')
    
    return list_of_strings
```

```python
table = pd.DataFrame()

filter = ['(U)' not in x for x in df['model']]

table['Model'] = df[filter]['model']

# table['Mean Accuracy'] = plus_minus(df['mean_accuracy'],df['std_accuracy'])
# table['Median Accuracy'] = [get_right_dec_figs_in_string(x,2) for x in np.round(df['median_accuracy'],2)]
table['Mean Balanced Accuracy [%]'] = plus_minus(df[filter]['mean_balanced_accuracy'],
                                             df[filter]['std_balanced_accuracy'])
# table['Median Balanced Accuracy'] = [get_right_dec_figs_in_string(x,2) for x in np.round(df['median_balanced_accuracy'],2)]

table.T #.style.hide(axis="index")
```

```python
# for class-wise accuracies
df_classwise = pd.read_csv('class_wise_accuracy.csv')
table_lm = pd.DataFrame()

comp_classes = [x for x in df_classwise.columns if all(y not in x for y in ['model', 'std', 'median'])]
comp_classes_lm = [x for x in comp_classes if 'phytochemical' not in x]
comp_classes_ru = [x for x in comp_classes if x not in ['lignin','tannin']]

idx_wanted = [i for i, model in enumerate(df_classwise['model']) if all(x not in model for x in ['(U)', 'MSCC', 'R-U'])]
table_lm['Model'] = df_classwise['model'].loc[idx_wanted]

for cclass in comp_classes_lm:
    table_lm[cclass] = plus_minus(df_classwise.loc[idx_wanted][cclass],
                               df_classwise.loc[idx_wanted][f'{cclass}_std'])

table_lm.T
```

```python
table_ru = pd.DataFrame()
idx_wanted = [i for i, model in enumerate(df_classwise['model']) if any(x in model for x in ['R-U','MSCC']) and '(U)' not in model]
table_ru['Model'] = df_classwise['model'].loc[idx_wanted]
for cclass in comp_classes_ru:
    table_ru[cclass] = plus_minus(df_classwise.loc[idx_wanted][cclass],
                                  df_classwise.loc[idx_wanted][f'{cclass}_std'])
table_ru.T
```
