```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('model_performance_comparison.csv')
df.columns
```

```python
def get_dec_places(x:float):
    string = str(x)
    string = string.split('.')

    for i in range(len(string[1])):
        if string[1][i] != '0':
            if int(string[1][i+1]) >= 5 and string[1][i]=='9':
                return i
            else: return i + 1

def get_right_dec_figs_in_string(n,dec_places):
    dec_figs = len(str(n).split('.')[1])
    if dec_figs < dec_places:
        return str(n) + ''.join(['0']*(dec_places-dec_figs))
    else:
        return str(n)

def plus_minus(mean,std):
    std_dec_places = [get_dec_places(x) for x in std]

    return [f'{get_right_dec_figs_in_string(np.round(x,z),z)} ± {get_right_dec_figs_in_string(np.round(y,z),z)}'\
             for x, y, z in zip(mean,std,std_dec_places)]
```

```python
table = pd.DataFrame()

filter = ['(U)' not in x for x in df['model']]

table['Model'] = df[filter]['model']

# table['Mean Accuracy'] = plus_minus(df['mean_accuracy'],df['std_accuracy'])
# table['Median Accuracy'] = [get_right_dec_figs_in_string(x,2) for x in np.round(df['median_accuracy'],2)]
table['Mean Balanced Accuracy'] = plus_minus(df[filter]['mean_balanced_accuracy'],
                                             df[filter]['std_balanced_accuracy'])
# table['Median Balanced Accuracy'] = [get_right_dec_figs_in_string(x,2) for x in np.round(df['median_balanced_accuracy'],2)]

table.style.hide(axis="index")
```
